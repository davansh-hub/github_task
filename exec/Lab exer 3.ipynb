```python
import numpy as np
A = np.array([[5, 2, 3],
              [6, 1, 3],
              [2, 4, 1]])
D = np.array([[4, -7, 4],[5 ,6 ,-1],[-4, 2, 5]])
AD = np.matmul(A, D)
DA = np.matmul(D, A)
print("AD:", AD)
print("DA:", DA)
print(np.array_equal(AD, DA))
```

```python
import numpy as np
B = np.array([[5, 2],
              [6, 1]])
det_B = np.linalg.det(B)
print("Determinant of B:", det_B)

inv_B = np.linalg.inv(B)
print("Inverse of B:", inv_B)
```

```python
import numpy as np
D = np.array([[4, -7, 4],[5 ,6 ,-1],[-4, 2, 5]])


inv_D = np.linalg.inv(D)
D_times_inv_D = np.matmul(D, inv_D)

inv_D_times_D = np.matmul(inv_D, D)

print("D * inv_D:")
print(D_times_inv_D)
print("inv_D * D:")
print(inv_D_times_D)

# we can observe this is identity matrix with diagonal 1
```
