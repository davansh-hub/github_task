# Medium Article Scraper

# URL: https://medium.com/@subashgandyer/papa-what-is-a-neural-network-c5e5cc427c7


````python
import os
import requests
import re
import sys
from bs4 import BeautifulSoup
````

````python
def get_page(url):
    """
    Get the HTML source of a Medium article
    
    Args:
        url (str): URL of the Medium article
    
    Returns:
        BeautifulSoup: Parsed HTML content
    """
    # Validate URL format
    if not re.match(r'https?://medium.com/', url):
        print('Please enter a valid website, or make sure it is a medium article')
        return None
    
    # Make HTTP request
    res = requests.get(url)
    res.raise_for_status()
    
    # Parse HTML content
    soup = BeautifulSoup(res.text, 'html.parser')
    return soup
````

### Function to clean HTML tags from text


````python
def clean(text):
    """
    Remove HTML tags and replace some with specific strings
    
    Args:
        text (str): Text containing HTML tags
    
    Returns:
        str: Cleaned text without HTML tags
    """
    rep = {"<br>": "\n", "<br/>": "\n", "<li>": "\n"}
    rep = dict((re.escape(k), v) for k, v in rep.items()) 
    pattern = re.compile("|".join(rep.keys()))
    text = pattern.sub(lambda m: rep[re.escape(m.group(0))], text)
    text = re.sub(r'\<(.*?)\>', '', text)
    return text
````

### Function to extract text content from the article


````python
def collect_text(soup, url):
    """
    Extract text content from parsed HTML
    
    Args:
        soup (BeautifulSoup): Parsed HTML content
        url (str): Original URL of the article
    
    Returns:
        str: Extracted text content
    """
    text = f'url: {url}\n\n'
    para_text = soup.find_all('p')
    print(f"Found {len(para_text)} paragraphs")
    
    for para in para_text:
        text += f"{para.text}\n\n"
    
    return text
````

### Function to save the extracted text to a file


````python
def save_file(text, url):
    """
    Save extracted text to a file in the scraped_articles directory
    
    Args:
        text (str): Text content to save
        url (str): Original URL to generate filename
    """
    # Create directory if it doesn't exist
    if not os.path.exists('./scraped_articles'):
        os.mkdir('./scraped_articles')
    
    # Generate filename from URL
    name = url.split("/")[-1]
    print(f"Filename: {name}")
    fname = f'scraped_articles/{name}.txt'
    
    # Write text to file
    with open(fname, 'w', encoding='utf-8') as f:
        f.write(text)
    
    print(f'File saved in directory {fname}')
    return fname
````

## Main Execution

Run the scraper by providing a Medium article URL


````python
# Get URL from user input
url = input("Enter url of a medium article: ")

# Process the article
try:
    # Get HTML content
    soup = get_page(url)
    
    if soup is not None:
        # Extract text content
        text = collect_text(soup, url)
        
        # Save to file
        filename = save_file(text, url)
        
        print("\n✅ Article successfully scraped and saved!")
        print(f"📁 File location: {filename}")
    else:
        print("❌ Failed to process the URL")
        
except Exception as e:
    print(f"❌ An error occurred: {str(e)}")
````

## Example Usage

You can also run the scraper programmatically without user input:


````python
# Example: Scrape a specific article programmatically
example_url = "https://medium.com/@subashgandyer/papa-what-is-a-neural-network-c5e5cc427c7"

print(f"Scraping article: {example_url}")

try:
    soup = get_page(example_url)
    
    if soup is not None:
        text = collect_text(soup, example_url)
        filename = save_file(text, example_url)
        
        print("\n✅ Example article successfully scraped!")
        print(f"📁 File location: {filename}")
        
        # Show first 500 characters of extracted text
        print("\n📄 Preview of extracted text:")
        print("-" * 50)
        print(text[:500] + "..." if len(text) > 500 else text)
    else:
        print("❌ Failed to process the example URL")
        
except Exception as e:
    print(f"❌ An error occurred: {str(e)}")
````
